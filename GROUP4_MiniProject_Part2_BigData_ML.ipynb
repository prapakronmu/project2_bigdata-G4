## 1. ปัญหาและคำถามที่สนใจ

**ปัญหา/คำถามที่สนใจคืออะไร**

1. **การวิเคราะห์รูปแบบความล่าช้าของเที่ยวบิน (Flight Delay Patterns):** ปัจจัยด้านสายการบิน สนามบินต้นทางและปลายทาง เส้นทางบิน และช่วงเวลาเดินทาง มีความสัมพันธ์กับโอกาสเกิดความล่าช้าของเที่ยวบินอย่างไร และมีรูปแบบความล่าช้าที่แตกต่างกันในแต่ละกลุ่มหรือไม่?

2. **การวิเคราะห์ความสัมพันธ์และการสะสมของความล่าช้า (Delay Relationships & Cascading Effects):** ความล่าช้าของเที่ยวบินมีรูปแบบที่อาจสะท้อนการสะสมหรือส่งผลต่อเที่ยวบินอื่นในระบบตารางบินอย่างไร โดยพิจารณาจากข้อมูลที่มีอยู่ และมีข้อจำกัดอะไรบ้างในการสรุปความสัมพันธ์เหล่านี้เป็นสาเหตุของความล่าช้า?

3. **การพัฒนาโมเดลทำนายและป้องกัน Data Leakage (Predictive Modeling & Target Leakage):** สามารถใช้ข้อมูลที่ทราบก่อนเวลาออกเดินทางตามกำหนดเพื่อทำนายว่าเที่ยวบินจะล่าช้าตั้งแต่ 15 นาทีขึ้นไปได้แม่นยำเพียงใด และโมเดล Machine Learning ใดให้ประสิทธิภาพเหมาะสมที่สุด โดยต้องเลือกตัวแปรอย่างระมัดระวังเพื่อป้องกัน Target Leakage และประเมินผลด้วยการแบ่งข้อมูลตามลำดับเวลา?

**ปัญหา/คำถามนี้มีความสำคัญอย่างไร และใครจะนำผลไปใช้**

ผลการศึกษาช่วยให้เข้าใจรูปแบบความล่าช้าและประเมินความเสี่ยงก่อนออกเดินทาง โดยผู้ที่อาจนำผลไปใช้ ได้แก่

- **สายการบินและสนามบิน:** วางแผนตารางบินและจัดสรรทรัพยากร
- **ผู้โดยสาร:** ใช้ประกอบการวางแผนเดินทาง
- **นักวิเคราะห์และนักวิจัย:** ศึกษารูปแบบความล่าช้าและประเมินประสิทธิภาพของโมเดล

ทั้งนี้ ความสัมพันธ์ของข้อมูลไม่สามารถยืนยันสาเหตุของความล่าช้าได้โดยตรง

**เป้าหมาย (target) และประเภทงาน (classification / regression / unsupervised)**

#### เป้าหมายของการทำนาย (Target): DEP_DEL15 ซึ่งระบุว่าเที่ยวบินออกเดินทางล่าช้าตั้งแต่ 15 นาทีขึ้นไปหรือไม่

1 = เที่ยวบินออกเดินทางล่าช้าตั้งแต่ 15 นาทีขึ้นไป

0 = เที่ยวบินออกเดินทางล่าช้าน้อยกว่า 15 นาที หรือไม่ล่าช้า

ประเภทงาน: Supervised Learning — Binary Classification เนื่องจากโมเดลต้องจำแนกเที่ยวบินออกเป็น 2 กลุ่ม ได้แก่ เที่ยวบินที่ล่าช้าตั้งแต่ 15 นาทีขึ้นไป และเที่ยวบินที่ไม่เข้าเกณฑ์ดังกล่าว

ข้อมูลที่ใช้ทำนาย: เลือกเฉพาะตัวแปรที่ทราบได้ก่อนเวลาออกเดินทางตามกำหนด เช่น สายการบิน สนามบินต้นทางและปลายทาง วันที่เดินทาง เวลาออกเดินทางตามกำหนด และระยะทาง โดยไม่นำข้อมูลที่เกิดขึ้นหลังเวลาออกเดินทาง เช่น เวลาล่าช้าจริง เวลานำเครื่องขึ้นจากหลุมจอด หรือเวลาที่เครื่องบินขึ้นบิน มาเป็นตัวแปรทำนาย เพราะอาจทำให้เกิด Data Leakage

การประเมินโมเดล: เปรียบเทียบโมเดลอย่างน้อย 3 วิธี เช่น Logistic Regression, Decision Tree และ Random Forest โดยแบ่งข้อมูลตามลำดับเวลา และใช้ตัวชี้วัด เช่น Precision, Recall, F1-score และ ROC-AUC เพื่อพิจารณาประสิทธิภาพของโมเดล

```python
import os
import shutil

output_dir = "/content/drive/MyDrive/Project_2Datasets/Processed"
os.makedirs(output_dir, exist_ok=True)

output_path = os.path.join(
    output_dir, "flights_2024.csv"
)

with open(output_path, "w", encoding="utf-8", newline="") as out:
    for i, path in enumerate(files):
        with open(path, "r", encoding="utf-8-sig", newline="") as src:
            if i > 0:
                next(src)  # ข้ามหัวตารางของไฟล์ที่ 2–12

            shutil.copyfileobj(src, out, length=1024 * 1024)

print("รวมไฟล์เรียบร้อย!")
print("บันทึกที่:", output_path)
```

```python
import pandas as pd

missing_counts = None
target_counts = {}
total_rows = 0
unnamed_nonmissing = 0

for chunk in pd.read_csv(
    output_path,
    chunksize=200_000,
    low_memory=False
):
    total_rows += len(chunk)

    # นับค่าว่างในทุกคอลัมน์
    chunk_missing = chunk.isna().sum()

    if missing_counts is None:
        missing_counts = chunk_missing
    else:
        missing_counts += chunk_missing

    # นับค่าของตัวแปรเป้าหมาย
    target_counts_chunk = chunk["DepDel15"].value_counts(dropna=False)

    for label, count in target_counts_chunk.items():
        target_counts[label] = target_counts.get(label, 0) + int(count)

    # ตรวจคอลัมน์ Unnamed
    if "Unnamed: 109" in chunk.columns:
        unnamed_nonmissing += int(
            chunk["Unnamed: 109"].notna().sum()
        )

print(f"จำนวนแถวทั้งหมด: {total_rows:,}")

print("\n=== จำนวนค่าว่างแต่ละคอลัมน์ ===")
print(missing_counts[missing_counts > 0].sort_values(ascending=False))

print("\n=== สัดส่วน Target: DepDel15 ===")
print(pd.Series(target_counts, name="จำนวนแถว"))

print("\n=== ตรวจคอลัมน์ Unnamed: 109 ===")
print("จำนวนค่าที่ไม่ว่าง:", unnamed_nonmissing)
```

```python
import pandas as pd

target_0 = 0
target_1 = 0
target_missing = 0
target_other = 0

for chunk in pd.read_csv(
    output_path,
    usecols=["DepDel15"],
    chunksize=200_000,
    low_memory=False
):
    s = chunk["DepDel15"]

    target_0 += int((s == 0).sum())
    target_1 += int((s == 1).sum())
    target_missing += int(s.isna().sum())
    target_other += int((s.notna() & ~s.isin([0, 1])).sum())

print("=== สรุป Target: DepDel15 ===")
print(f"ไม่ล่าช้าถึง 15 นาที (0): {target_0:,}")
print(f"ล่าช้าตั้งแต่ 15 นาทีขึ้นไป (1): {target_1:,}")
print(f"ไม่มีค่า Target: {target_missing:,}")
print(f"ค่าอื่นที่ไม่ใช่ 0 หรือ 1: {target_other:,}")
print(f"รวมทั้งหมด: {target_0 + target_1 + target_missing + target_other:,}")
```

```python
import pandas as pd

missing_target_counts = {}

for chunk in pd.read_csv(
    output_path,
    usecols=["DepDel15", "Cancelled", "Diverted"],
    chunksize=200_000,
    low_memory=False
):
    # แปลงค่าว่างให้เป็นหมวดเดียวกัน
    for col in ["DepDel15", "Cancelled", "Diverted"]:
        chunk[col] = (
            chunk[col]
            .astype("string")
            .fillna("Missing")
            .str.strip()
        )

    # เลือกเฉพาะแถวที่ Target ว่าง
    missing_rows = chunk[chunk["DepDel15"] == "Missing"]

    counts = missing_rows.groupby(
        ["Cancelled", "Diverted"]
    ).size()

    for key, count in counts.items():
        missing_target_counts[key] = (
            missing_target_counts.get(key, 0) + int(count)
        )

result = pd.DataFrame(
    [
        {
            "Cancelled": key[0],
            "Diverted": key[1],
            "จำนวนแถว": count
        }
        for key, count in missing_target_counts.items()
    ]
)

print("=== สถานะของแถวที่ไม่มี DepDel15 ===")
print(result.to_string(index=False))
print(
    "\nรวมแถวที่ไม่มี Target:",
    f"{result['จำนวนแถว'].sum():,}"
)
```

```python
import os
import pandas as pd

output_dir = "/content/drive/MyDrive/Project_2Datasets/Processed"
os.makedirs(output_dir, exist_ok=True)

clean_path = os.path.join(output_dir, "flights_2024_clean.csv")
model_path = os.path.join(output_dir, "flights_2024_model.csv")

# ตัวนับสำหรับตรวจสอบผลหลังทำความสะอาด
total_rows = 0
clean_rows = 0
model_rows = 0
target_counts = {0: 0, 1: 0}
cancelled_rows = 0

# อ่านข้อมูลทีละส่วน
for i, chunk in enumerate(
    pd.read_csv(output_path, chunksize=200_000, low_memory=False)
):
    total_rows += len(chunk)

    # 1. ตัดคอลัมน์ที่ว่างทั้งหมด
    if "Unnamed: 109" in chunk.columns:
        chunk = chunk.drop(columns=["Unnamed: 109"])

    # 2. บันทึกชุดข้อมูลสะอาดสำหรับ EDA
    chunk.to_csv(
        clean_path,
        mode="w" if i == 0 else "a",
        header=(i == 0),
        index=False
    )
    clean_rows += len(chunk)

    # 3. เตรียมชุดข้อมูลสำหรับโมเดล
    cancelled_rows += int((chunk["Cancelled"] == 1).sum())

    model_chunk = chunk[
        (chunk["Cancelled"] == 0) &
        (chunk["DepDel15"].notna())
    ].copy()

    # นับ Target ก่อนบันทึก
    for label in [0, 1]:
        target_counts[label] += int(
            (model_chunk["DepDel15"] == label).sum()
        )

    model_chunk.to_csv(
        model_path,
        mode="w" if i == 0 else "a",
        header=(i == 0),
        index=False
    )
    model_rows += len(model_chunk)

print("=== ผล Data Cleaning ===")
print(f"จำนวนแถวต้นฉบับ: {total_rows:,}")
print(f"จำนวนแถวในไฟล์ Clean สำหรับ EDA: {clean_rows:,}")
print(f"จำนวนเที่ยวบินที่ยกเลิกในข้อมูล: {cancelled_rows:,}")
print(f"จำนวนแถวในไฟล์ Model: {model_rows:,}")
print(f"Target 0: {target_counts[0]:,}")
print(f"Target 1: {target_counts[1]:,}")
print(f"รวม Target 0 และ 1: {sum(target_counts.values()):,}")

print("\nบันทึกไฟล์เรียบร้อย")
print("ไฟล์ Clean:", clean_path)
print("ไฟล์ Model:", model_path)
```
