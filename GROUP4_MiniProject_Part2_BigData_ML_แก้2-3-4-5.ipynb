**ปัญหา/คำถามนี้มีความสำคัญอย่างไร และใครจะนำผลไปใช้**

ผลการศึกษาช่วยให้เข้าใจรูปแบบความล่าช้าและประเมินความเสี่ยงก่อนออกเดินทาง โดยผู้ที่อาจนำผลไปใช้ ได้แก่

- **สายการบินและสนามบิน:** วางแผนตารางบินและจัดสรรทรัพยากร
- **ผู้โดยสาร:** ใช้ประกอบการวางแผนเดินทาง
- **นักวิเคราะห์และนักวิจัย:** ศึกษารูปแบบความล่าช้าและประเมินประสิทธิภาพของโมเดล

ทั้งนี้ ความสัมพันธ์ของข้อมูลไม่สามารถยืนยันสาเหตุของความล่าช้าได้โดยตรง

**เป้าหมาย (target) และประเภทงาน (classification / regression / unsupervised)**

#### เป้าหมายของการทำนาย (Target): DEP_DEL15 ซึ่งระบุว่าเที่ยวบินออกเดินทางล่าช้าตั้งแต่ 15 นาทีขึ้นไปหรือไม่

1 = เที่ยวบินออกเดินทางล่าช้าตั้งแต่ 15 นาทีขึ้นไป

0 = เที่ยวบินออกเดินทางล่าช้าน้อยกว่า 15 นาที หรือไม่ล่าช้า

ประเภทงาน: Supervised Learning — Binary Classification เนื่องจากโมเดลต้องจำแนกเที่ยวบินออกเป็น 2 กลุ่ม ได้แก่ เที่ยวบินที่ล่าช้าตั้งแต่ 15 นาทีขึ้นไป และเที่ยวบินที่ไม่เข้าเกณฑ์ดังกล่าว

ข้อมูลที่ใช้ทำนาย: เลือกเฉพาะตัวแปรที่ทราบได้ก่อนเวลาออกเดินทางตามกำหนด เช่น สายการบิน สนามบินต้นทางและปลายทาง วันที่เดินทาง เวลาออกเดินทางตามกำหนด และระยะทาง โดยไม่นำข้อมูลที่เกิดขึ้นหลังเวลาออกเดินทาง เช่น เวลาล่าช้าจริง เวลานำเครื่องขึ้นจากหลุมจอด หรือเวลาที่เครื่องบินขึ้นบิน มาเป็นตัวแปรทำนาย เพราะอาจทำให้เกิด Data Leakage

การประเมินโมเดล: เปรียบเทียบโมเดลอย่างน้อย 3 วิธี เช่น Logistic Regression, Decision Tree และ Random Forest โดยแบ่งข้อมูลตามลำดับเวลา และใช้ตัวชี้วัด เช่น Precision, Recall, F1-score และ ROC-AUC เพื่อพิจารณาประสิทธิภาพของโมเดล

ขั้นตอนนี้ดาวน์โหลดหรือค้นหาไฟล์ต้นทาง 12 เดือน ตรวจสอบจำนวนไฟล์และชื่อคอลัมน์ แล้วอ่านเฉพาะ 11 คอลัมน์ที่ใช้ในการศึกษา ก่อนรวมเป็น df_raw โดยยังไม่ลบแถวซ้ำ ไม่คัด Outliers และไม่เปลี่ยนแปลง Target ในขั้นโหลดข้อมูล

### 3.1 เตรียมไลบรารีและกำหนดตำแหน่งไฟล์


```python
import os
import sys
import glob
import zipfile
import shutil
import importlib.util
import subprocess

# ติดตั้ง gdown เฉพาะเมื่อ environment ยังไม่มีไลบรารีนี้
if importlib.util.find_spec("gdown") is None:
    subprocess.check_call([
        sys.executable, "-m", "pip", "install", "-q", "gdown"
    ])
import gdown

# ZIP ที่เก็บสำเนาข้อมูล BTS ปี 2024 ของกลุ่ม
GOOGLE_DRIVE_ZIP_ID = "1urPqDoHP0e3rhzaQkPrhDDVjX3kpc7C7"
ZIP_FILENAME = "flights_2024.zip"
EXTRACT_DIR = "./csv_files"
```

### 3.2 ดาวน์โหลดและแตกไฟล์ข้อมูลเมื่อยังมีไฟล์ไม่ครบ


```python
def find_csv_files():
    return sorted(glob.glob(os.path.join(EXTRACT_DIR, "**", "*.csv"), recursive=True))

csv_files = find_csv_files()

# ถ้าไฟล์ไม่ครบ 12 เดือน ให้ล้างโฟลเดอร์ที่แตกไว้ก่อนหน้าเพื่อป้องกันไฟล์เก่าปะปน
if len(csv_files) != 12:
    print("พบไฟล์ CSV ไม่ครบ 12 ไฟล์ กำลังดาวน์โหลดชุดข้อมูลใหม่...")

    if os.path.isdir(EXTRACT_DIR):
        shutil.rmtree(EXTRACT_DIR)
    os.makedirs(EXTRACT_DIR, exist_ok=True)

    if os.path.exists(ZIP_FILENAME):
        os.remove(ZIP_FILENAME)

    url = f"https://drive.google.com/uc?id={GOOGLE_DRIVE_ZIP_ID}"
    downloaded_file = gdown.download(url, ZIP_FILENAME, quiet=False)

    if downloaded_file is None or not os.path.isfile(ZIP_FILENAME):
        raise RuntimeError(
            "ดาวน์โหลดไม่สำเร็จ กรุณาตรวจสอบว่าไฟล์ Google Drive เปิดให้ผู้มีลิงก์เข้าถึงได้"
        )

    with zipfile.ZipFile(ZIP_FILENAME, "r") as zip_ref:
        bad_file = zip_ref.testzip()
        if bad_file is not None:
            raise RuntimeError(f"ไฟล์ ZIP เสียหาย: {bad_file}")
        zip_ref.extractall(EXTRACT_DIR)

    csv_files = find_csv_files()

print("จำนวนไฟล์ CSV ที่พบ:", len(csv_files))
if len(csv_files) != 12:
    raise RuntimeError(
        f"ต้องพบไฟล์ CSV 12 ไฟล์ แต่พบ {len(csv_files)} ไฟล์ กรุณาตรวจสอบ ZIP ต้นทาง"
    )
```

### 3.3 ตรวจสอบจำนวนไฟล์


```python
csv_files = find_csv_files()

print("จำนวนไฟล์ CSV:", len(csv_files))
assert len(csv_files) == 12, (
    f"พบไฟล์ CSV {len(csv_files)} ไฟล์ ต้องมี 12 ไฟล์สำหรับข้อมูลครบทั้งปี"
)

print("โฟลเดอร์ข้อมูล:", os.path.abspath(EXTRACT_DIR))
print("รายชื่อไฟล์:")
for file in csv_files:
    print("-", os.path.basename(file))
```

## 4. Data Quality Assessment
ตรวจและรายงานเป็นตัวเลข: ชนิดข้อมูล, missing values, ข้อมูลซ้ำ, ค่าผิดปกติ/เป็นไปไม่ได้, ความไม่สอดคล้องของข้อมูล

```python
# 4.1 ภาพรวมข้อมูล ประเภทข้อมูล และหน่วยความจำ

import pandas as pd
import numpy as np

print("=" * 60)
print("4.1 ภาพรวมและชนิดข้อมูล")
print("=" * 60)

total_rows = len(df_raw)
total_cols = len(df_raw.columns)
memory_mb = df_raw.memory_usage(deep=True).sum() / (1024 ** 2)

print(f"จำนวนแถว: {total_rows:,}")
print(f"จำนวนคอลัมน์: {total_cols:,}")
print(f"หน่วยความจำที่ใช้: {memory_mb:.2f} MB")

print("\nชนิดข้อมูลแต่ละคอลัมน์:")
print(df_raw.dtypes)

print("\nจำนวนคอลัมน์แยกตามชนิดข้อมูล:")
print(df_raw.dtypes.value_counts())
```

```python
# 4.2 ตรวจสอบ Missing Values

print("=" * 60)
print("4.2 ตรวจสอบข้อมูลที่หายไป")
print("=" * 60)

missing_df = pd.DataFrame({
    "Missing Count": df_raw.isna().sum(),
    "Missing Percent (%)": (
        df_raw.isna().mean() * 100
    ).round(2)
})

active_missing = missing_df[
    missing_df["Missing Count"] > 0
].sort_values(
    "Missing Count",
    ascending=False
)

print(f"จำนวนช่องข้อมูลที่หายทั้งหมด: {df_raw.isna().sum().sum():,}")

if len(active_missing) > 0:
    print("\nคอลัมน์ที่พบข้อมูลหาย:")
    print(active_missing.to_string())
else:
    print("ไม่พบข้อมูลที่หายไป")
```

```python
# 4.3 ตรวจสอบแถวที่เหมือนกันทุกคอลัมน์ (ยังไม่ลบ)

print("=" * 60)
print("4.3 ตรวจสอบข้อมูลซ้ำ")
print("=" * 60)

duplicate_count = int(df_raw.duplicated().sum())
duplicate_percent = (
    duplicate_count / len(df_raw) * 100
    if len(df_raw) > 0 else 0
)

print(f"จำนวนแถวทั้งหมด: {len(df_raw):,}")
print(f"จำนวนแถวซ้ำ: {duplicate_count:,}")
print(f"เปอร์เซ็นต์แถวซ้ำ: {duplicate_percent:.4f}%")

if duplicate_count > 0:
    print("\nตัวอย่างแถวซ้ำ 5 แถว:")
    print(
        df_raw.loc[
            df_raw.duplicated(keep=False)
        ].head(5).to_string(index=False)
    )
else:
    print("ไม่พบแถวซ้ำ")
```

```python
# 4.4 Invalid Values

print("=" * 60)
print("4.4 ตรวจสอบค่าที่ผิดเงื่อนไข")
print("=" * 60)

invalid_checks = {
    "Month": ~df_raw["Month"].between(1, 12),

    "DayofMonth": ~df_raw["DayofMonth"].between(1, 31),

    "DayOfWeek": ~df_raw["DayOfWeek"].between(1, 7),

    "CRSDepTime": (
        (df_raw["CRSDepTime"] < 0) |
        (df_raw["CRSDepTime"] > 2400) |
        (
            (df_raw["CRSDepTime"] != 2400) &
            (
                (df_raw["CRSDepTime"] // 100 > 23) |
                (df_raw["CRSDepTime"] % 100 > 59)
            )
        )
    ),

    "Distance": (
        df_raw["Distance"].notna() &
        (df_raw["Distance"] <= 0)
    ),

    "Cancelled": (
        df_raw["Cancelled"].notna() &
        ~df_raw["Cancelled"].isin([0, 1])
    ),

    "DepDel15": (
        df_raw["DepDel15"].notna() &
        ~df_raw["DepDel15"].isin([0, 1])
    )
}

print("จำนวนแถวที่ผิดเงื่อนไขในแต่ละคอลัมน์:")

for col, invalid_mask in invalid_checks.items():
    print(f"{col}: {int(invalid_mask.sum()):,} แถว")

# แสดงรายละเอียดแถวที่มีเวลาไม่ถูกต้อง
invalid_time = invalid_checks["CRSDepTime"]

if invalid_time.any():
    print("\nตัวอย่างแถวที่เวลาไม่ถูกต้อง:")
    print(
        df_raw.loc[
            invalid_time,
            [
                "FlightDate",
                "Reporting_Airline",
                "Origin",
                "Dest",
                "CRSDepTime"
            ]
        ].head(10).to_string(index=False)
    )
```

```python
# 4.5 ตรวจค่าที่อยู่นอกเกณฑ์ IQR ของ Distance เพื่อการตรวจสอบเพิ่มเติมเท่านั้น
# การอยู่นอกเกณฑ์ IQR ไม่ได้แปลว่าเป็นข้อมูลผิด และจะไม่ลบแถวจากผลนี้โดยอัตโนมัติ
q1_distance = df_raw["Distance"].quantile(0.25)
q3_distance = df_raw["Distance"].quantile(0.75)
iqr_distance = q3_distance - q1_distance
lower_distance = q1_distance - 1.5 * iqr_distance
upper_distance = q3_distance + 1.5 * iqr_distance
potential_distance_outliers = df_raw["Distance"].lt(lower_distance) | df_raw["Distance"].gt(upper_distance)

print(f"Q1 ของ Distance: {q1_distance:.2f}")
print(f"Q3 ของ Distance: {q3_distance:.2f}")
print(f"ขอบเขต IQR: {lower_distance:.2f} ถึง {upper_distance:.2f}")
print(f"จำนวนค่าที่อยู่นอกขอบเขต IQR: {int(potential_distance_outliers.sum()):,}")
print("ให้ถือเป็นค่าที่ควรตรวจสอบเพิ่มเติม ไม่ใช่หลักฐานว่าค่าเหล่านี้ผิด")
```

```python
# 4.5 Data Inconsistency: Extra Whitespace

print("=" * 60)
print("4.5 ตรวจสอบความไม่สอดคล้องของข้อมูล")
print("=" * 60)

cat_cols = df_raw.select_dtypes(
    include=["object", "category", "string"]
).columns

inconsistency_summary = {}

for col in cat_cols:
    text_values = df_raw[col].astype("string")

    has_extra_spaces = text_values.str.contains(
        r"^\s+|\s+$",
        regex=True,
        na=False
    )

    inconsistency_summary[col] = int(
        has_extra_spaces.sum()
    )

inconsistency_df = pd.Series(
    inconsistency_summary,
    name="Rows with Extra Whitespace"
)

print("จำนวนแถวที่มีช่องว่างเกินในแต่ละคอลัมน์:")

print(inconsistency_df.to_string())

print(
    "\nผลรวมจำนวนแถวที่พบช่องว่างเกิน:",
    f"{inconsistency_df.sum():,}"
)

if inconsistency_df.sum() == 0:
    print("ไม่พบช่องว่างส่วนเกินในคอลัมน์ข้อความ")
```

## 5. Data Cleaning
อธิบายทุกการตัดสินใจ **พร้อมเหตุผล** และรายงานจำนวนแถวที่เปลี่ยนแปลงในแต่ละขั้น

| ขั้นตอน | สิ่งที่ทำ | เหตุผล | แถวคงเหลือ |
|---|---|---|---|
| ... | ... | ... | ... |

```python
# 5.1 เริ่มต้น Data Cleaning และสร้าง Audit Trail
import pandas as pd
import numpy as np

print("=" * 80)
print("5. DATA CLEANING")
print("=" * 80)

initial_rows = len(df_raw)
df_clean = df_raw.copy()
cleaning_log = []

def log_cleaning_step(step, action, reason, rows_before, rows_removed, rows_after):
    cleaning_log.append({
        "ขั้นตอน": step,
        "สิ่งที่ทำ": action,
        "เหตุผล": reason,
        "แถวก่อน": int(rows_before),
        "แถวที่ลบ": int(rows_removed),
        "แถวคงเหลือ": int(rows_after),
    })

log_cleaning_step(
    "0. ข้อมูลเริ่มต้น",
    "คัดลอก df_raw เป็น df_clean โดยไม่แก้ข้อมูลต้นฉบับ",
    "เก็บข้อมูลดิบไว้เพื่อให้ตรวจสอบย้อนกลับได้",
    initial_rows, 0, len(df_clean)
)
print(f"จำนวนแถวเริ่มต้น: {initial_rows:,}")
print(f"จำนวนคอลัมน์: {len(df_clean.columns):,}")
```

```python
# 5.2 ตรวจสอบแถวซ้ำ แต่ยังไม่ลบอัตโนมัติ
rows_before = len(df_clean)
duplicate_extra_rows = int(df_clean.duplicated().sum())
duplicate_group_rows = int(df_clean.duplicated(keep=False).sum())

# ไม่ใช้ drop_duplicates() ในขั้นนี้ เพราะ 11 คอลัมน์ที่เลือกมายังไม่มี Flight ID
# สำหรับยืนยันว่าแถวเหมือนกันหมายถึงเที่ยวบินเดียวกันจริง
log_cleaning_step(
    "1. ตรวจสอบข้อมูลซ้ำ",
    f"พบแถวที่ซ้ำเกินจากการเก็บแถวแรก {duplicate_extra_rows:,} แถว; ยังไม่ลบ",
    "ข้อมูลที่เหมือนกันในคอลัมน์ที่เลือกอาจเป็นคนละเที่ยวบิน ต้องตรวจสอบกับรหัสเที่ยวบิน/ข้อมูลต้นทางก่อนลบ",
    rows_before, 0, len(df_clean)
)
print(f"แถวซ้ำเกินที่พบ: {duplicate_extra_rows:,}")
print(f"แถวทั้งหมดที่อยู่ในกลุ่มซ้ำ: {duplicate_group_rows:,}")
print(f"แถวที่ลบในขั้นนี้: 0")
print(f"แถวคงเหลือ: {len(df_clean):,}")
```

```python
# 5.3 ลบเฉพาะแถวที่ทุกคอลัมน์ว่างทั้งหมด
rows_before = len(df_clean)
empty_rows = int(df_clean.isna().all(axis=1).sum())
df_clean = df_clean.dropna(how="all").copy()
rows_removed = rows_before - len(df_clean)

log_cleaning_step(
    "2. จัดการแถวว่างทั้งหมด",
    f"ลบแถวที่ทุกคอลัมน์ว่าง {rows_removed:,} แถว",
    "แถวที่ไม่มีข้อมูลเลยไม่สามารถนำไปวิเคราะห์ได้; ไม่ลบแถวที่มีข้อมูลบางคอลัมน์หายไปโดยอัตโนมัติ",
    rows_before, rows_removed, len(df_clean)
)
print(f"แถวว่างทั้งหมดที่พบ: {empty_rows:,}")
print(f"แถวที่ลบ: {rows_removed:,}")
print(f"แถวคงเหลือ: {len(df_clean):,}")
```

```python
# 5.4 แปลง FlightDate เป็น datetime และบันทึกค่าที่แปลงไม่ได้
rows_before = len(df_clean)
missing_date_before = int(df_clean["FlightDate"].isna().sum())
parsed_flight_date = pd.to_datetime(df_clean["FlightDate"], errors="coerce")
missing_date_after = int(parsed_flight_date.isna().sum())
new_parse_failures = max(0, missing_date_after - missing_date_before)
df_clean["FlightDate"] = parsed_flight_date

log_cleaning_step(
    "3. แปลงชนิดข้อมูลวันที่",
    f"แปลง FlightDate เป็น datetime; ค่าว่าง/แปลงไม่ได้หลังแปลง {missing_date_after:,} แถว (พบการแปลงไม่ได้เพิ่ม {new_parse_failures:,})",
    "รองรับการจัดกลุ่มตามวันและตรวจสอบแนวโน้มตามเวลา; ไม่ลบแถววันที่ผิดรูปแบบโดยไม่ตรวจสอบก่อน",
    rows_before, 0, len(df_clean)
)
print("ชนิดข้อมูล FlightDate หลังแปลง:", df_clean["FlightDate"].dtype)
print(f"ค่าว่างของวันที่ก่อนแปลง: {missing_date_before:,}")
print(f"ค่าว่าง/แปลงไม่ได้หลังแปลง: {missing_date_after:,}")
print(f"จำนวนแถวคงเหลือ: {len(df_clean):,}")
```

```python
# 5.5 ตัดช่องว่างส่วนเกินในคอลัมน์ข้อความ
rows_before = len(df_clean)
text_cols = df_clean.select_dtypes(include=["object", "string", "category"]).columns.tolist()
whitespace_fixed = 0

for col in text_cols:
    before_text = df_clean[col].astype("string")
    after_text = before_text.str.strip()
    changed = before_text.ne(after_text).fillna(False)
    whitespace_fixed += int(changed.sum())
    df_clean[col] = after_text

log_cleaning_step(
    "4. จัดการช่องว่างข้อความ",
    f"ตัดช่องว่างหัว/ท้ายในคอลัมน์ข้อความ; เปลี่ยนแปลง {whitespace_fixed:,} ค่า",
    "ป้องกันข้อความที่มีช่องว่างต่างกันถูกจัดเป็นคนละหมวดหมู่",
    rows_before, 0, len(df_clean)
)
print(f"คอลัมน์ข้อความที่ตรวจ: {len(text_cols)}")
print(f"ค่าที่เปลี่ยนแปลง: {whitespace_fixed:,}")
print(f"แถวคงเหลือ: {len(df_clean):,}")
```

```python
# 5.6 ประเมิน Outliers โดยไม่ลบอัตโนมัติ
rows_before = len(df_clean)
log_cleaning_step(
    "5. ประเมินค่าผิดปกติ",
    "ไม่ลบแถวด้วย IQR อัตโนมัติ; ใช้ผลจาก Part 4 เพื่อคัดเลือกค่าที่ควรตรวจสอบ",
    "ระยะทางที่ห่างจากกลุ่มส่วนใหญ่อาจเป็นเส้นทางบินจริง; ต้องตรวจสอบบริบทหรือแหล่งข้อมูลก่อนระบุว่าเป็นข้อผิดพลาด",
    rows_before, 0, len(df_clean)
)
print("ไม่มีการลบแถวด้วย IQR ในขั้นนี้")
print("ข้อมูล DepDel15 ที่ว่างจะเก็บไว้ใน df_clean และค่อยคัดออกจากชุดสำหรับโมเดลในขั้น Modeling")
print(f"แถวคงเหลือ: {len(df_clean):,}")
```

### การตัดสินใจและเหตุผล

**1. การจัดการข้อมูลซ้ำ (Handle Duplicates)**

- **สิ่งที่ทำ:** ตรวจสอบและลบแถวที่มีข้อมูลซ้ำกันทุกคอลัมน์ โดยเก็บแถวแรกไว้
- **เหตุผล:** ข้อมูลซ้ำอาจทำให้การนับจำนวนเที่ยวบินและการสรุปผลทางสถิติคลาดเคลื่อน จึงลบแถวที่ซ้ำกันเพื่อป้องกันปัญหา Double Counting จากการตรวจสอบพบข้อมูลซ้ำ 1,377 แถว

**2. การจัดการแถวที่ว่างทั้งหมด (Handle Completely Empty Rows)**

- **สิ่งที่ทำ:** ตรวจสอบและลบเฉพาะแถวที่ทุกคอลัมน์เป็นค่าว่าง
- **เหตุผล:** แถวที่ไม่มีข้อมูลเลยไม่สามารถนำไปใช้วิเคราะห์ได้ อย่างไรก็ตาม จะไม่ลบแถวที่มีข้อมูลบางคอลัมน์หายไปโดยอัตโนมัติ เพราะข้อมูลส่วนอื่นอาจยังมีประโยชน์ต่อการวิเคราะห์

**3. การแปลงชนิดข้อมูลวันที่ (Convert Date Data Type)**

- **สิ่งที่ทำ:** แปลงคอลัมน์ `FlightDate` จากข้อความให้เป็นชนิดข้อมูลวันที่และเวลา (`datetime`)
- **เหตุผล:** ช่วยให้สามารถจัดเรียงข้อมูล แยกข้อมูลตามช่วงเวลา และวิเคราะห์แนวโน้มจำนวนเที่ยวบินหรือความล่าช้าตามวันที่ได้สะดวกขึ้น โดยตรวจสอบวันที่ที่ว่างหรือแปลงไม่ได้เพิ่มเติม

**4. การจัดการความไม่สอดคล้องของข้อความ (Handle Inconsistent Text)**

- **สิ่งที่ทำ:** ตัดช่องว่างส่วนเกินบริเวณหัวและท้ายข้อความในคอลัมน์ประเภทข้อความ
- **เหตุผล:** ป้องกันปัญหาข้อความที่เหมือนกันแต่มีช่องว่างต่างกันจนถูกมองเป็นคนละค่า ซึ่งอาจกระทบต่อการจัดกลุ่มและการกรองข้อมูล จากการตรวจสอบเบื้องต้นใน Part 4 ไม่พบช่องว่างส่วนเกินในคอลัมน์ข้อความที่ตรวจ

**5. การตรวจสอบค่าผิดปกติ (Assess Outliers)**

- **สิ่งที่ทำ:** ไม่ลบข้อมูลด้วยวิธี IQR โดยอัตโนมัติ แต่พิจารณาความสมเหตุสมผลของข้อมูลก่อนตัดสินใจลบ
- **เหตุผล:** ข้อมูลเที่ยวบินบางรายการอาจมีค่าระยะทางหรือเวลาแตกต่างจากข้อมูลส่วนใหญ่ แต่ยังเป็นข้อมูลจริงได้ การใช้ IQR กับทุกคอลัมน์ตัวเลขโดยไม่แยกประเภทตัวแปรอาจทำให้ข้อมูลที่ถูกต้องถูกลบออกจำนวนมาก ดังนั้นจึงควรตรวจสอบค่าที่ผิดเงื่อนไขและบริบทของข้อมูลก่อน
